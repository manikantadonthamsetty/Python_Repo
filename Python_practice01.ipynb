```python

#Practice of Python for loop with else statement
files = ['file1.csv', 'file2.csv', 'file3.csv','file4.csv']

for file in files:
    if not file.endswith('.csv'):
        print(f"{file} is not a CSV file.")
        break
else:
    print("All files are CSV files.")
```

```python
#Practice of Python for loop with else statement
file_list = ['report.csv','data.xlsx','summary.docx','report.csv','data.csv']

for file in file_list:
    if file_list.count(file) > 1:
        print(f"{file} is not unique.")
        break
else:
    print('All are Unique')
```

```python
#Nested Loops

for x in range(3):
    for y in range(2):
        for z in range(1):
            print (f"({x}, {y}, {z})")
    
```
